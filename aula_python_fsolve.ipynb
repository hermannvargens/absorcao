## 2. Tomando Decisões: `if`, `elif` e `else`

Nem sempre o programa segue uma linha reta. Às vezes ele precisa tomar caminhos diferentes dependendo de uma condição.

* `if` (se): avalia a primeira condição.
* `elif` (senão se): avalia uma alternativa caso a anterior seja falsa.
* `else` (senão): executado caso nenhuma condição anterior seja atendida.

```python
# Simulando a medição de uma temperatura
temperatura_reator = 105.0

if temperatura_reator > 100.0:
    print("⚠ Alerta: Temperatura acima do limite seguro!")
elif temperatura_reator >= 60.0:
    print("✅ Operação normal e estável.")
else:
    print("❄️ Temperatura abaixo da faixa de operação.")
```

## 3. Repetições Controladas: `for` e `while`

O computador é excelente em fazer a mesma tarefa milhares de vezes sem cansar.

* **`for`**: Usamos quando sabemos quantas vezes queremos repetir algo (ou queremos percorrer uma sequência).
* **`while`**: Repete um bloco **enquanto** uma condição continuar sendo verdadeira.

```python
# Exemplo com 'for': contar de 1 até 5
print("--- Teste com FOR ---")
for passo in range(1, 6):
    print(f"Contando: {passo}")

# Exemplo com 'while': encher um tanque até o nível desejado
print("\n--- Teste com WHILE ---")
nivel_tanque = 0
limite = 4

while nivel_tanque < limite:
    nivel_tanque += 1  # o mesmo que: nivel_tanque = nivel_tanque + 1
    print(f"Nível atual: {nivel_tanque} m")

print("Tanque cheio!")
```

## 4. Criando Nossas Próprias Ferramentas: Funções (`def`)

Uma função é um bloco de código reaproveitável. Você fornece uma entrada e ela devolve uma resposta.

```python
# Criando uma função que calcula a diferença entre x² e 5
# Se a resposta for 0, significa que achamos a raiz quadrada de 5!
def residuo(x):
    resultado = x**2 - 5
    return resultado

# Testando a função com valores manuais
print("Se x = 2:", residuo(2))  # 2² - 5 = -1 (faltou)
print("Se x = 3:", residuo(3))  # 3² - 5 = +4 (passou)
```

## 5. Resolvendo "Na Mão": Método da Bisseção

Nosso desafio: **qual é o valor exato de $x$ para que $x^2 - 5 = 0$?**

* Vimos que para $x = 2$, o erro é **negativo** (ficou abaixo de 5).
* Para $x = 3$, o erro é **positivo** (passou de 5).
* Portanto, a resposta **precisa** estar entre 2 e 3!

Vamos usar um laço `while` com `if`/`else` para testar o ponto médio e cortar o intervalo pela metade a cada rodada.

```python
# Limites onde sabemos que a raiz está confinada
limite_baixo = 2.0
limite_alto = 3.0
tolerancia = 0.000000001  # Critério de parada: erro muito próximo de zero

while True:
    # 1. Chuta exatamente no meio do caminho
    chute = (limite_baixo + limite_alto) / 2
    erro = residuo(chute)

    print(f"Chute = {chute:.5f} | Erro = {erro:+.5f}")

    # 2. O erro já é pequeno o suficiente?
    if abs(erro) < tolerancia:
        print(f"\n🎉 Raiz encontrada pelo método manual: {chute:.5f}")
        break

    # 3. Ajusta o intervalo baseado no sinal do erro
    if erro > 0:
        limite_alto = chute      # Passou, reduz o teto
    else:
        limite_baixo = chute     # Faltou, sobe o piso

```

## 6. Usando a Função `fsolve`

O método manual acima funcionou bem, mas precisou de várias iterações e nós tivemos que fornecer dois limites conhecidos.

Em engenharia, usamos bibliotecas especializadas. O `fsolve` (da biblioteca `scipy`) faz exatamente essa busca, mas utiliza algoritmos muito mais inteligentes baseados na inclinação da curva, convergindo em pouquíssimos passos.

```python

# Importando a função fsolve
from scipy.optimize import fsolve

# 1. Definimos a equação igualada a zero (já declaramos a função 'residuo' acima)
# def residuo(x):
#     return x**2 - 5

# 2. Fornecemos apenas um 'chute inicial' aproximado
chute_inicial = 2.0

# 3. Executamos o fsolve
solucao = fsolve(residuo, x0=chute_inicial)

print(f"Chute inicial: {chute_inicial}")
print(f"Solução encontrada pelo fsolve: {solucao[0]:.5f}")

# Prova real
print(f"Prova real: ({solucao[0]:.5f})² = {solucao[0]**2:.5f}")
```
